```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv("C:\\Users\\Piyush Verma\\Desktop\\Clean_Dataset.csv")
print("Initial shape:", df.shape)
print(df.head())
print(df.info())
```

# 2. BASIC CLEANING

```python
# Drop unnecessary columns
df.drop(columns=["Sno", "flight"], errors="ignore", inplace=True)

print("\nAfter dropping ID columns, shape:", df.shape)

# Check missing values BEFORE cleaning
print("\nMissing values per column (before cleaning):")
print(df.isna().sum())

# ✅ Step 1: Remove completely empty rows
df = df.dropna(how='all')

# ✅ Step 2: Remove rows where target is missing
df = df.dropna(subset=["price"])

# ✅ Step 3: Reset index
df.reset_index(drop=True, inplace=True)

# Final check
print("\nShape after cleaning:", df.shape)

print("\nNaN in target after cleaning:", df["price"].isnull().sum())

print("\nRemaining missing values:")
print(df.isna().sum())
```

# 3. FEATURES & TARGET

```python
TARGET_COL = "price"

X = df.drop(columns=[TARGET_COL])
y = df[TARGET_COL]

print("\nFeature columns:", X.columns.tolist())
print("Target:", TARGET_COL)
```

# 4. SIMPLE ENCODING

```python
X_encoded = pd.get_dummies(X, drop_first=True)

print("\nShape after encoding:", X_encoded.shape)
print("Encoded columns example:", X_encoded.columns[:10])
```

# 5. TRAIN–TEST SPLIT

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42
)

print("\nTrain shape:", X_train.shape)
print("Test shape:", X_test.shape)
```

# 6. LINEAR REGRESSION (FIRST MODEL)

```python
from sklearn.ensemble import HistGradientBoostingRegressor

model = HistGradientBoostingRegressor()
model.fit(X_train, y_train)
```

```python
lr = LinearRegression()

print("\nTraining Linear Regression...")
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)

mae_lr = mean_absolute_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mean_squared_error(y_test, y_pred_lr))
r2_lr = r2_score(y_test, y_pred_lr)

print("\n=== LINEAR REGRESSION PERFORMANCE ===")
print(f"MAE  : {mae_lr:.2f}")
print(f"RMSE : {rmse_lr:.2f}")
print(f"R²   : {r2_lr:.4f}")
```

# 7. RANDOM FOREST REGRESSOR (SECOND MODEL)

```python
rf = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

print("\nTraining Random Forest...")
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)

mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

print("\n=== RANDOM FOREST PERFORMANCE ===")
print(f"MAE  : {mae_rf:.2f}")
print(f"RMSE : {rmse_rf:.2f}")
print(f"R²   : {r2_rf:.4f}")
```

# Comparison

```python
print("\n=== MODEL COMPARISON ===")
print(f"Linear Regression  RMSE: {rmse_lr:.2f} | R²: {r2_lr:.4f}")
print(f"Random Forest      RMSE: {rmse_rf:.2f} | R²: {r2_rf:.4f}")
```

# 8. ACTUAL vs PREDICTED PRICE COMPARISON (Scatter Plot)

####  Using Random Forest predictions

```python
y_test_pred = y_pred_rf

comparison_df = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_test_pred
})

print("\n=== Actual vs Predicted Price (First 10 Rows) ===")
print(comparison_df.head(10))

plt.figure(figsize=(7,6))
plt.scatter(y_test, y_pred_rf, alpha=0.25, s=18, color="royalblue")
plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()],
         "--", color="red", linewidth=2)
plt.title("Actual vs Predicted")
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.grid(alpha=0.3)
plt.show()
```

# 9. FEATURE IMPORTANCE

```python
feat_imp = pd.DataFrame({
    "feature": X_encoded.columns,
    "importance": rf.feature_importances_
})
feat_imp["orig"] = feat_imp["feature"].apply(lambda x: x.split("_")[0])
agg_imp = feat_imp.groupby("orig")["importance"].sum().sort_values(ascending=False)

print("\nFeature importance by original feature:")
print(agg_imp)


plt.bar(agg_imp.index, agg_imp.values)
plt.xticks(rotation=45)
plt.title("Feature Importance (Original Features)")
plt.tight_layout()
plt.show()
```

# 10. TRAIN FINAL RANDOM FOREST ON FULL DATA FOR ANALYSIS

```python
rf_full = RandomForestRegressor(
    n_estimators=200, random_state=42, n_jobs=-1
)

rf_full.fit(X_encoded, y)
df_analysis = df.copy()
df_analysis["predicted_price"] = rf_full.predict(X_encoded)
```

```python

max_days = df_analysis["days_left"].max()
base_bins = [0, 3, 7, 14, 30, 60, 120]
bins = [b for b in base_bins if b < max_days] + [max_days + 1]
labels = [f"{bins[i]}-{bins[i+1]-1}" for i in range(len(bins)-1)]

df_analysis["days_left_bin"] = pd.cut(
    df_analysis["days_left"], bins=bins, labels=labels, right=False
)

print("\nEarly booking price trend:")
print(df_analysis.groupby("days_left_bin")["predicted_price"].mean())

plt.plot(labels, df_analysis.groupby("days_left_bin")["predicted_price"].mean(), marker="o")
plt.xticks(rotation=45)
plt.title("Predicted Price vs Days Left")
plt.show()
```

# STOPS INCREASE COST

```python
print("\nStops vs predicted price:")
print(df_analysis.groupby("stops")["predicted_price"].mean())

grp = df_analysis.groupby("stops")["predicted_price"].mean()
plt.bar(grp.index, grp.values, color=["orange","blue","green"])
plt.title("Price vs Stops")
plt.show()
```

# DURATION AFFECT PRICE? 

```python
max_dur = df_analysis["duration"].max()
dur_bins = [0,2,4,8, max_dur + 0.01]
dur_labels = ["0-2", "2-4", "4-8", "8+"]

df_analysis["duration_bin"] = pd.cut(
    df_analysis["duration"], bins=dur_bins, labels=dur_labels, right=False
)

print("\nDuration vs predicted price:")
print(df_analysis.groupby("duration_bin")["predicted_price"].mean())

grp = df_analysis.groupby("duration_bin")["predicted_price"].mean()
plt.bar(grp.index, grp.values, color=["purple","brown","pink","cyan"])
plt.title("Price vs Duration")
plt.show()
```

#  WHICH AIRLINE IS MOST EXPENSIVE? 

```python
airline_summary = df_analysis.groupby("airline")["predicted_price"].mean().sort_values(ascending=False)

print("\nAirlines ranked by predicted price:")
print(airline_summary.head(10))

top = airline_summary[:10]
plt.bar(top.index, top.values, color=plt.cm.Set3(range(10)))
plt.xticks(rotation=45)
plt.title("Airline Price Ranking")
plt.show()
```
